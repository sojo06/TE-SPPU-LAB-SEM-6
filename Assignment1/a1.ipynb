# Formating date to YYYY-MM-DD

```python
 df['Premiere'] = pd.to_datetime(df['Premiere'], errors='coerce') #formatting date 
 df.head()
```

# Replace Imdb score with mean
IMDB Score is a continuous numerical variable that usually lies within a limited range (0–10) 
and tends to be approximately normally distributed.

The mean represents the average rating of all movies.
Replacing missing IMDB scores with the mean helps maintain the overall rating distribution.
Mean replacement does not introduce bias when the data has no extreme outliers.

```python
df['IMDB Score'] = df['IMDB Score'].fillna(df['IMDB Score'].mean())
```

# Replace Runtime  with median 
Runtime is also a numerical variable, but it often contains outliers (very short or very long movies).

A few extremely long movies can skew the mean.
The median represents the middle value and is not affected by outliers.
Using the median ensures that missing runtime values are replaced with a typical movie length.

```python
df['Runtime'] = df['Runtime'].fillna(df['Runtime'].median())
```

# Replace language with mode 
Language is a categorical (text-based) attribute, not numerical.

Mean and median cannot be applied to text data.
The mode represents the most frequently occurring language in the dataset.
Replacing missing language values with the mode preserves the most common language pattern in the data.

```python
most_common_language = df['Language'].mode()[0]
df['Language'] = df['Language'].fillna(most_common_language)
```
