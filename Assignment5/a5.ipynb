# Import necessary libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

# Load Inbuilt Titanic Dataset

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

# Check Dataset Information

```python
titanic.info()
```

# Data Preprocessing

```python
titanic.isnull().sum()
```

# Fill missing age values with median age


```python
titanic["age"]=titanic["age"].fillna(titanic["age"].median())

titanic.isnull().sum()
```

# Box Plot – Age Distribution by Gender & Survival
- Females have a higher survival rate across most age groups.
- Male passengers show a wider age spread among non-survivors.
- Younger passengers have higher chances of survival compared to older passengers.
- Outliers indicate presence of very young and very old passengers.
- Survival depends on both age and gender.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="sex", y="age", hue="survived", data=titanic)
plt.xlabel("Gender")
plt.ylabel("Age")
plt.title("Age Distribution by Gender and Survival Status")
plt.show()
```
